```python


import pandas as pd

df = pd.read_csv(
    '/content/Sample - Superstore.csv',
    encoding='latin1'
)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

df.head()

```

```python


print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())
```

```python



print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

print("\nBasic statistics:")
display(df.describe())
```

```python



# Create a separate copy for cleaning
df_clean = df.copy()

# Remove completely duplicated rows
df_clean = df_clean.drop_duplicates()

# Convert date columns, if they exist
for col in ['Order Date', 'Ship Date']:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col], errors='coerce'
        )

# Show the results
print("Original rows:", len(df))
print("Rows after removing duplicates:", len(df_clean))
print("Duplicate rows remaining:", df_clean.duplicated().sum())

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nCleaned data preview:")
display(df_clean.head())
```

```python



if 'Sales' in df_clean.columns:
    total_sales = df_clean['Sales'].sum()
    print(f"Total Sales: {total_sales:,.2f}")

if 'Profit' in df_clean.columns:
    total_profit = df_clean['Profit'].sum()
    print(f"Total Profit: {total_profit:,.2f}")

print("Total Orders:", len(df_clean))

if 'Order ID' in df_clean.columns:
    print("Unique Orders:", df_clean['Order ID'].nunique())
```

```python



if 'Category' in df_clean.columns and 'Sales' in df_clean.columns:

    category_sales = (
        df_clean.groupby('Category')['Sales']
        .sum()
        .sort_values(ascending=False)
    )

    print("Sales by Category:")
    display(category_sales)

    category_sales.plot(
        kind='bar',
        figsize=(8, 5),
        title='Total Sales by Product Category',
        xlabel='Product Category',
        ylabel='Total Sales',
        rot=0
    )

    plt.tight_layout()
    plt.show()

else:
    print("Category or Sales column not found.")
```

```python



# Make sure the Order Date column is in date format
df_clean['Order Date'] = pd.to_datetime(
    df_clean['Order Date'], errors='coerce'
)

# Remove rows where the order date or sales value is missing
monthly_data = df_clean.dropna(
    subset=['Order Date', 'Sales']
).copy()

# Group sales by month
monthly_data['Month'] = (
    monthly_data['Order Date'].dt.to_period('M')
)

monthly_sales = (
    monthly_data.groupby('Month')['Sales']
    .sum()
    .sort_index()
)

print("Monthly Sales:")
display(monthly_sales)

# Draw a line chart
monthly_sales.plot(
    kind='line',
    figsize=(12, 5),
    marker='o',
    title='Monthly Sales Trend',
    xlabel='Month',
    ylabel='Total Sales',
    grid=True
)

plt.tight_layout()
plt.show()
```

```python

# Step 12: Identify the highest- and lowest-sales months

best_month = monthly_sales.idxmax()
worst_month = monthly_sales.idxmin()

print("Highest Sales Month:", best_month)
print(f"Sales: {monthly_sales.max():,.2f}")

print("\nLowest Sales Month:", worst_month)
print(f"Sales: {monthly_sales.min():,.2f}")
```

```python



product_sales = (
    df_clean.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
)

top_10_products = product_sales.head(10)

print("Top 10 Products by Total Sales:")
display(top_10_products)

# Create a bar chart
top_10_products.sort_values().plot(
    kind='barh',
    figsize=(10, 6),
    title='Top 10 Products by Total Sales',
    xlabel='Total Sales',
    ylabel='Product Name'
)

plt.tight_layout()
plt.show()
```

```python



category_profit = (
    df_clean.groupby('Category')['Profit']
    .sum()
    .sort_values(ascending=False)
)

print("Total Profit by Category:")
display(category_profit)

# Create a bar chart
category_profit.plot(
    kind='bar',
    figsize=(8, 5),
    title='Total Profit by Product Category',
    xlabel='Product Category',
    ylabel='Total Profit',
    rot=0
)

plt.axhline(y=0, linewidth=1)
plt.tight_layout()
plt.show()
```

```python



category_summary = (
    df_clean.groupby('Category')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum')
    )
)

category_summary['Profit_Margin_%'] = (
    category_summary['Total_Profit']
    / category_summary['Total_Sales']
) * 100

display(
    category_summary.sort_values(
        by='Total_Profit',
        ascending=False
    ).round(2)
)
```

```python



region_summary = (
    df_clean.groupby('Region')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum')
    )
    .sort_values(by='Total_Sales', ascending=False)
)

print("Sales and Profit by Region:")
display(region_summary.round(2))

# Compare sales across regions
region_summary['Total_Sales'].plot(
    kind='bar',
    figsize=(8, 5),
    title='Total Sales by Region',
    xlabel='Region',
    ylabel='Total Sales',
    rot=0
)

plt.tight_layout()
plt.show()
```

```python



total_sales = df_clean['Sales'].sum()
total_profit = df_clean['Profit'].sum()
total_orders = df_clean['Order ID'].nunique()
total_products = df_clean['Product Name'].nunique()

profit_margin = (total_profit / total_sales) * 100

print("=" * 40)
print("       RETAIL SALES SUMMARY")
print("=" * 40)

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Unique Orders: {total_orders:,}")
print(f"Unique Products: {total_products:,}")
print(f"Overall Profit Margin: {profit_margin:.2f}%")

print("=" * 40)
```

```python



import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Prepare dashboard data
dashboard_df = df_clean.copy()

dashboard_df['Order Date'] = pd.to_datetime(
    dashboard_df['Order Date'], errors='coerce'
)

total_sales = dashboard_df['Sales'].sum()
total_profit = dashboard_df['Profit'].sum()
total_orders = dashboard_df['Order ID'].nunique()
total_products = dashboard_df['Product Name'].nunique()
profit_margin = (total_profit / total_sales) * 100

# Display dashboard title and KPIs
display(Markdown("# RETAIL SALES ANALYTICS DASHBOARD"))
display(Markdown(
    "**Business overview:** Sales, profit, products and trends"
))

display(Markdown(
    f"""
    | Key Performance Indicator | Value |
    |:--|--:|
    | Total Sales | {total_sales:,.2f} |
    | Total Profit | {total_profit:,.2f} |
    | Unique Orders | {total_orders:,} |
    | Unique Products | {total_products:,} |
    | Profit Margin | {profit_margin:.2f}% |
    """
))

# Create a 2-by-2 dashboard of charts
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle(
    "Retail Sales Performance Dashboard",
    fontsize=18,
    fontweight='bold'
)

# Chart 1: Sales by category
dashboard_df.groupby('Category')['Sales'].sum().sort_values(
    ascending=False
).plot(kind='bar', ax=axes[0, 0], title='Sales by Category')

axes[0, 0].set_ylabel('Total Sales')
axes[0, 0].tick_params(axis='x', rotation=0)

# Chart 2: Profit by category
dashboard_df.groupby('Category')['Profit'].sum().sort_values(
    ascending=False
).plot(kind='bar', ax=axes[0, 1], title='Profit by Category')

axes[0, 1].set_ylabel('Total Profit')
axes[0, 1].tick_params(axis='x', rotation=0)

# Chart 3: Monthly sales trend
monthly_sales_dashboard = (
    dashboard_df.dropna(subset=['Order Date', 'Sales'])
    .assign(
        Month=lambda x: x['Order Date'].dt.to_period('M')
    )
    .groupby('Month')['Sales'].sum()
    .sort_index()
)

axes[1, 0].plot(
    monthly_sales_dashboard.index.astype(str),
    monthly_sales_dashboard.values,
    marker='o'
)

axes[1, 0].set_title('Monthly Sales Trend')
axes[1, 0].set_ylabel('Total Sales')
axes[1, 0].tick_params(axis='x', rotation=60)

# Chart 4: Top 10 products by sales
top_products_dashboard = (
    dashboard_df.groupby('Product Name')['Sales'].sum()
    .nlargest(10).sort_values()
)

top_products_dashboard.plot(
    kind='barh', ax=axes[1, 1],
    title='Top 10 Products by Sales'
)

axes[1, 1].set_xlabel('Total Sales')

plt.tight_layout()
plt.show()
```

```python



monthly_sales_dashboard.plot(
    kind='line',
    figsize=(9, 4),
    marker='o',
    title='Monthly Sales Trend',
    xlabel='Month',
    ylabel='Total Sales',
    grid=True
)

ax = plt.gca()


ax.set_xticks(range(0, len(monthly_sales_dashboard), 6))
ax.set_xticklabels(
    [str(monthly_sales_dashboard.index[i])
     for i in range(0, len(monthly_sales_dashboard), 6)],
    rotation=45,
    ha='right'
)

plt.tight_layout()
plt.show()
```

```python


region_summary = (
    df_clean.groupby('Region')
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

display(region_summary.round(2))

region_summary[['Total_Sales', 'Total_Profit']].plot(
    kind='bar',
    figsize=(9, 5),
    title='Sales and Profit by Region'
)

plt.xlabel('Region')
plt.ylabel('Amount')
plt.xticks(rotation=0)

plt.tight_layout()

# Save the dashboard image
fig.savefig(
    '/content/retail_sales_dashboard.png',
    dpi=200,
    bbox_inches='tight'
)

plt.show()

print("Dashboard image saved successfully!")

```

## Business Insights and Recommendations

### 1. Overall Business Performance
The dataset records total sales of 2,297,200.86 and total profit of 286,397.02 across 5,009 unique orders. The overall profit margin is approximately 12.47%.

**Recommendation:** Monitor sales and profit together to understand overall business performance.

### 2. Product Category Performance
Technology generated the highest total profit of 145,454.95, followed by Office Supplies at 122,490.80. Furniture generated a profit of 18,451.27.

**Recommendation:** Investigate the factors behind category-level differences in profitability.

### 3. Furniture Profit Margin
Furniture has a profit margin of only 2.49%, compared with 17.40% for Technology and 17.04% for Office Supplies.

**Recommendation:** Examine Furniture product-level profits, discounts, and costs to identify opportunities to improve margins.

### 4. Monthly Sales Trends
November 2017 recorded the highest monthly sales of 118,447.82, while February 2014 recorded the lowest at 4,519.89.

**Recommendation:** Investigate seasonal patterns and other factors that may explain monthly sales fluctuations.

### 5. Product-Level Performance
The Canon imageCLASS 2200 Advanced Copier was the highest-selling product by total sales value in the dataset.

**Recommendation:** Examine its profit margin and sales volume before making inventory or promotional decisions.
